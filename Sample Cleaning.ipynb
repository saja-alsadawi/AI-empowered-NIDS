```python
import pandas as pd
import numpy as np

# Load the sample dataset
file_path = r'C:\Users\user\CSE-CIC-IDS2018-Full\sample.csv'
df = pd.read_csv(file_path, low_memory=False)

print(" Initial shape:", df.shape)

# Remove duplicates 
before_dupes = df.shape[0]
df = df.drop_duplicates()
after_dupes = df.shape[0]
print(f" Removed {before_dupes - after_dupes} duplicate rows")

# Convert all features to numeric exclude 'Label'
label = df['Label']
df = df.drop(columns=['Label', 'Timestamp'], errors='ignore')

# Convert features to numeric 
df = df.apply(pd.to_numeric, errors='coerce')
df['Label'] = label

# Fill in missing values with column median 
missing_before = df.isnull().sum().sum()
print(f" Missing values before filling: {missing_before}")

df = df.fillna(df.median(numeric_only=True))

missing_after = df.isnull().sum().sum()
print(f" Missing values after filling: {missing_after}")

# Final info
print(" Final cleaned shape:", df.shape)
print("\n Data type summary:")
print(df.dtypes.value_counts())


output_path = r'C:\Users\user\CSE-CIC-IDS2018-Full\clean_sample.csv'
df.to_csv(output_path, index=False)
print(f"Cleaned sample saved to: {output_path}")
```
