```python
import pandas as pd

# Load sample
df = pd.read_csv(r'C:\Users\user\CSE-CIC-IDS2018-Full\sample.csv')

# Shape
print("Shape:", df.shape)

# Class distribution
class_counts = df['Label'].value_counts()
class_percent = df['Label'].value_counts(normalize=True) * 100
print("\n Class distribution:")
print(pd.DataFrame({'Count': class_counts, 'Percentage': class_percent.round(4)}))

# Columns
print("\n Column count:", len(df.columns))
print("\n First 5 columns:", df.columns[:5].tolist())
print("\n Last 5 columns:", df.columns[-5:].tolist())

# Missing values
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print("\n Missing values:")
print(missing)

# Duplicates
num_duplicates = df.duplicated().sum()
print(f"\n Duplicate rows: {num_duplicates}")

# Data types
print("\n📦 Data types summary:")
print(df.dtypes.value_counts())

```
